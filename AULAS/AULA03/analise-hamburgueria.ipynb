**17 - Criação dos principais gráficos**

```python
sns.countplot(data=df, y="produto")
plt.title("Produtos mais vendidos")
```

```python
df["data"] = pd.to_datetime(df["data"])
df["mes"] = df["data"].dt.month
df.groupby("mes")["total"].sum()
```

**Avaliação por canal**

```python
df.groupby("canal")["avaliacao"].mean()
```

**Tempo de espera por canal**

```python
df.groupby("canal")["tempo_espera_min"].mean()
```

```python
faturamento_por_produto = df.groupby("produto")["total"].sum().reset_index()
sns.barplot(data = faturamento_por_produto, x = "total", y = "produto")
plt.title("Faturamento por produto")
plt.show()
```

```python
faturamento_por_dia_semana = df.groupby("dia_semana")["total"].sum().reset_index()
sns.barplot(data = faturamento_por_dia_semana, x = "total", y = "dia_semana")
plt.title("Faturamento por dia da semana")
plt.show()
```

```python
numeros = df[["quantidade","preco_unitario","tempo_espera_min","avaliacao","total","mes"]]
sns.heatmap(numeros.corr().round(2), annot = True)
```
